```python
import numpy  as np
```

```python
a=np.array([1,2,3])
b=np.array([3,4,5])
```

```python
a+b #ADD ELEMENTS OF TWO LISTS AT SAME INDEX
```

```python
a-b #SUBTRACT ELEMENTS OF TWO LISTS AT SAME INDEX
```

```python
a*b #MULTIPLY ELEMENTS OF TWO LISTS AT SAME INDEX
```

```python
a/b #DIVIDE ELEMENTS OF TWO LISTS AT SAME INDEX
```

```python
a//b #FLOOR DIVIDE ELEMENTS OF TWO LISTS AT SAME INDEX
```

```python
a**b #POWER ELEMENTS OF TWO LISTS AT SAME INDEX
```
